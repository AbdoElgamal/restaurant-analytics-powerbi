```sql
CREATE TABLE restaurant_final
using Delta
AS
SELECT order_id,
       cast(order_date as timestamp) as order_date,
       hour,
       category,
       item_name,
       price,
       quantity,
       discount,
       total_amount,
       branch,
       payment_method,
       order_type,
       customer_id,
       rating,
       is_weekend
FROM restaurant_csv

UNION ALL

SELECT order_id,
       order_date,
       hour,
       category,
       item_name,
       price,
       quantity,
       discount,
       total_amount,
       branch,
       payment_method,
       order_type,
       customer_id,
       rating,
       is_weekend
FROM restaurant_json
```

```sql
SELECT count(*) FROM restaurant_final;
```

```sql
ALTER TABLE restaurant_final
ADD COLUMNS (Year INT, Month INT, Day_Name STRING)
```

```sql
SELECT * FROM restaurant_final
```

```sql
UPDATE restaurant_final
SET Year = YEAR(order_date),
    Month = MONTH(order_date),
    Day_Name = date_format(order_date, 'EEEE');

SELECT * FROM restaurant_final;
```
